```python
import cv2
import os
from ultralytics import YOLO

# === CONFIGURATION ===
model_path = "Cricket_bat_toe_detection/yolov8l_custom2/weights/best.pt"
video_path = "Image/3.mp4"
output_dir = "Output"
os.makedirs(output_dir, exist_ok=True)

# === LOAD MODEL ===
model = YOLO(model_path)

# === LOAD VIDEO ===
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print(f"❌ Failed to open video: {video_path}")
    exit()

# === CLASS COLORS ===
# Format: class_id : (B, G, R)
class_colors = {
    0: (0, 255, 0),     # Toe - Green
    1: (255, 0, 0),     # Bat - Blue
    2: (0, 0, 255),     # Optional: False Object or Gloves - Red
}

frame_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # === INFERENCE ===
    results = model(frame, conf=0.25)[0]

    # === DRAW DETECTIONS ===
    for box in results.boxes:
        cls_id = int(box.cls[0])
        conf = float(box.conf[0])
        x1, y1, x2, y2 = map(int, box.xyxy[0])

        label = f"{model.names[cls_id]} {conf:.2f}"
        color = class_colors.get(cls_id, (255, 255, 255))  # White default if class not defined

        # Draw rectangle and label
        cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
        cv2.putText(frame, label, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

    # === SAVE FRAME ===
    out_path = os.path.join(output_dir, f"frame_{frame_count:04d}.jpg")
    success = cv2.imwrite(out_path, frame)
    if not success:
        print(f"❌ Failed to save frame: {out_path}")
    frame_count += 1

cap.release()
print(f"✅ Detection complete. {frame_count} frames saved in '{output_dir}'")
```

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import os

# Load your trained model
model = YOLO('Cricket_bat_toe_detection/yolov8l_custom2/weights/best.pt')

# Evaluate model on validation set
metrics = model.val(data='My-First-Project-4\data.yaml', split='val', conf=0.25)

# Plot and save the confusion matrix
fig = metrics.plot_confusion_matrix()
save_path = 'output_frames/confusion_matrix.png'

# Create output directory if not exists
os.makedirs(os.path.dirname(save_path), exist_ok=True)

# Save the figure
fig.savefig(save_path)
print(f"✅ Confusion matrix saved at: {save_path}")

# Show it
plt.show()
```
